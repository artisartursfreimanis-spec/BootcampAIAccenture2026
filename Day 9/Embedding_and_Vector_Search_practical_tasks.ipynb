#**Create local vector embeddings using sentens-transformer python library**

##**GOAL: to embed text sentences and perform semantic searches using your own Python code.**

There are many pre-trained embedding models available on Hugging Face that you can use to create vector embeddings.
Sentence Transformers (SBERT) is a library that makes it easy to use these models for vector embedding.

Use pip  to install  'sentence_transformers' library  and import  'SentenceTransformer model loader' from this library.

```python
#%pip install --no-compile sentence-transformers
from sentence_transformers import SentenceTransformer
```

Load the 'paraphrase-MiniLM-L6-v2' model  from HuggingFace resource  using the  SentenceTransformer( *model-name* )  and store the reference to the model object in the 'model' variable

```python
model = SentenceTransformer("paraphrase-MiniLM-L6-v2")
```

After loading the model, call the 'encode()' method on the model object to create a vector representation of a specific text sentence. Use your own text string  as the parameter.

```python
sentence = "My dog loves running on the beach every weekend."
embedding = model.encode(sentence)
embedding
```

Create vector representations for several text sentences. Place the text strings in a list and use this list as an argument. Use 8-10 sentences of 20-25 words each.  Call the 'encode()' method on the model object with the list of sentences as an argument.

```python

sentences_list = [
    "Dogs are loyal companions that love long walks in the park, playing fetch with their owners, and resting beside them every evening.",
    "My puppy gets very excited every morning when I grab his leash, because he knows we are going outside for a walk.",
    "Astronomers use powerful telescopes to study distant galaxies, hoping to understand how stars and planets formed billions of years ago in space.",
    "The rocket launched successfully this morning, carrying a new satellite into orbit to help scientists observe changes in the Earth climate.",
    "Regular exercise such as running, swimming or cycling strengthens the heart, improves mood and helps people stay healthy as they grow older.",
    "Eating plenty of fresh vegetables and fruit every day gives the body important vitamins and keeps the immune system strong and working well.",
    "Machine learning models learn patterns from large amounts of data and can then make predictions about new examples they have never seen.",
    "Neural networks are a type of artificial intelligence inspired by the human brain, made of layers of connected neurons that process information.",
    "Latvia is a small country on the Baltic Sea, known for its beautiful forests, sandy beaches and the historic old town of Riga.",
    "Every summer many tourists visit the Baltic coast to enjoy the beaches, eat local food and explore the charming small towns nearby.",
]

embeddings = model.encode(sentences_list)
print(embeddings.shape)
```

#**Definition of semantic textual similaritye**

Import 'util' module from sentence_transformers library.

```python
from sentence_transformers import util
```

You can calculate the cosine similarity of the vector representations of our sentences using the 'cos_sim()' function from the util module.
Example: sim = util.cos_sim(embedding_1, embedding_2). Calculate the cosine similarity for any two sentences from your list.


```python
sim_same_topic = util.cos_sim(embeddings[0], embeddings[1])
sim_diff_topic = util.cos_sim(embeddings[0], embeddings[2])
print("Dogs vs puppy:", sim_same_topic.item())
print("Dogs vs galaxies:", sim_diff_topic.item())
```

Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
def cos_similarity_calculation(text, sentences, sentence_embeddings):
    text_embedding = model.encode(text)
    scores = util.cos_sim(text_embedding, sentence_embeddings)[0]
    results = sorted(zip(scores.tolist(), sentences), reverse=True)
    for score, sentence in results:
        print(f"{score:.3f}  {sentence}")
    return results

results = cos_similarity_calculation("What is the best food for staying healthy?", sentences_list, embeddings)
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
import numpy as np

def cosine_similarity_batch(vector, batch):
    vector = np.asarray(vector, dtype=float)
    batch = np.asarray(batch, dtype=float)
    return (batch @ vector) / (np.linalg.norm(batch, axis=1) * np.linalg.norm(vector))

v = np.array([1, 0])
batch = np.array([[1, 0], [0, 1], [-1, 0], [1, 1]])
print(cosine_similarity_batch(v, batch))

query = model.encode("Tell me about space exploration")
numpy_scores = cosine_similarity_batch(query, embeddings)
util_scores = util.cos_sim(query, embeddings)[0].numpy()
print("NumPy:", np.round(numpy_scores, 3))
print("util: ", np.round(util_scores, 3))
print("Most similar:", sentences_list[np.argmax(numpy_scores)])
```
